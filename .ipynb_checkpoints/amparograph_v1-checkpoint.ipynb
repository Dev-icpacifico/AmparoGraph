```python
class State(TypedDict):
    graph_state: str
```

#### Creación del agente supervisor
*Utilizará LLM con salida estructurada para elegir el próximo nodo de trabajador o el procesamiento de finalización.*

```python
members = ["amparo", "buker", "planner", "reglamento", "search_web"]
options = members +["FINISH"]

system_promt =(
    "Eres un supervisor de RRHH que trabaja para las empresas constructora del mar II e inmobiliaria costanera pacifico"
    f" Y eres el encargado de gestionar las conversaciones y solicitudes entre los siguientes miembros:{members}."
    "Ante la siguiente solicitud del usuario,responda con el miembro para que actue acontinuación. "
    "Cada miembro realizará una tarea y responderá con sus resultados y el estado. Al finalizar responda con FINISH"
)
```

```python
class Router(TypedDict):
    """Worker to route to next. If no workers needed, route to FINISH."""
    next: Literal[options]
```

```python
llm =  ChatOpenAI(model= "gpt 4.0")
```

#### Creación de los Nodos o agentes

```python
def supervisor_node(state: State) -> Command[Literal[members, "__end__"]]:
    messages = [
        {"role": "system", "content": system_promt},
    ]+state["messages"]
    response = llm.with_structured_output(Router).invoke(messages)
    goto = response["next"]
    if goto == "FINISH":
        goto =END
    
    return Command(goto=goto, update={"next":goto})
```

```python
research_agent = create_react_agent(
    llm, tools=[tavily_tool], prompt="You are a researcher. DO NOT do any math."
)


def research_node(state: State) -> Command[Literal["supervisor"]]:
    result = research_agent.invoke(state)
    return Command(
        update={
            "messages": [
                HumanMessage(content=result["messages"][-1].content, name="researcher")
            ]
        },
        goto="supervisor",
    )
```
